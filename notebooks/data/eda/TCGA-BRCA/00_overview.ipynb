# TCGA-BRCA EDA: Overview and Navigation

This notebook provides an overview of the TCGA Breast Cancer (BRCA) dataset and demonstrates how to use the new `XenaCohort` class for easy data loading.

## Dataset Structure

The TCGA-BRCA dataset includes comprehensive multi-omics data:

### Available Data Types and Loading Methods

1. **Clinical Data** → `cohort.clinical()` ([01_clinical_analysis.ipynb](./01_clinical_analysis.ipynb))
   - Patient demographics
   - Tumor characteristics (stage, grade, TNM classification)
   - Treatment information
   - Sample metadata

2. **Survival Data** → `cohort.clinical()` ([02_survival_analysis.ipynb](./02_survival_analysis.ipynb))
   - Overall Survival (OS)
   - Progression-Free Interval (PFI)
   - Disease-Specific Survival (DSS)
   - Disease-Free Interval (DFI)

3. **PAM50 Molecular Subtypes** → `cohort.clinical()` ([03_pam50_subtype_analysis.ipynb](./03_pam50_subtype_analysis.ipynb))
   - Basal-like
   - HER2-enriched
   - Luminal A
   - Luminal B
   - Normal-like

4. **Gene Expression (RNA-seq)** → `cohort.mrna_seq()` ([04_gene_expression_analysis.ipynb](./04_gene_expression_analysis.ipynb))
   - TPM (Transcripts Per Million)
   - FPKM (Fragments Per Kilobase Million)
   - FPKM-UQ (Upper Quartile normalized)
   - Raw counts

5. **Copy Number Variation** → `cohort.cnv()`
   - Gene-level CNV (ASCAT2, ASCAT3, ASCAT-NGS)
   - Allele-specific CNV
   - Segment-level CNV

6. **DNA Methylation** → `cohort.methylation()`
   - Methylation450 (450K array)
   - Methylation27 (27K array)

7. **Somatic Mutations** → `cohort.mutation()`
   - Whole exome sequencing (WXS)

8. **Protein Expression (RPPA)** → `cohort.protein()`
   - Reverse Phase Protein Array

9. **miRNA Expression** → `cohort.mirna_seq()`
   - microRNA expression profiles

10. **ATAC-seq Peak Data** → `cohort.genomics()`
    - Chromatin accessibility

## Using XenaCohort in Your Analysis

All notebooks in this directory now use the `XenaCohort` class for simplified data loading. Instead of manually loading TSV files, you can now:

```python
from oncolearn.api.xenabrowser import XenaCohortBuilder

# Initialize
builder = XenaCohortBuilder()
cohort = builder.build_cohort("BRCA")

# Load any data type
data = cohort.clinical()  # or mrna_seq(), protein(), etc.
```

Benefits:
- ✓ Automatic merging of related datasets
- ✓ Source tracking via `_source_dataset` column
- ✓ Consistent interface across all data types
- ✓ Optional merging on sample IDs
