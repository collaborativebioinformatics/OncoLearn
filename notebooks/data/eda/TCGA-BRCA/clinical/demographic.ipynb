# TCGA-BRCA Comprehensive Demographic Analysis

This notebook provides comprehensive exploratory data analysis (EDA) of all 9 demographic columns using interactive Bokeh visualizations.

**Analysis Coverage:**
- Race and ethnicity distributions
- Gender analysis
- Age distributions (multiple representations)
- Vital status and mortality data
- Birth and death temporal patterns
- Cross-demographic comparisons

```python
# Import libraries
from oncolearn.api.xenabrowser import XenaCohortBuilder
import pandas as pd
import numpy as np
from pathlib import Path
# Bokeh imports (still needed for rendering)
from bokeh.plotting import show, output_notebook

# HoloViews imports
import holoviews as hv
from holoviews import opts, dim
hv.extension('bokeh')

# import warnings
# warnings.filterwarnings('ignore')

# Enable Bokeh in notebook
output_notebook()
```

## Load Clinical Data

Using the `XenaCohort` class to load all clinical datasets.

```python
# Build the cohort
builder = XenaCohortBuilder()
cohort = builder.build_cohort("BRCA")

# Load clinical data
clinical = cohort.clinical()
```

```python
gender_col = 'gender.demographic'
race_col = 'race.demographic'
vital_col = 'vital_status.demographic'
age_col = 'age_at_index.demographic'
yod_col = 'year_of_death.demographic'
```

## Extract Demographic Columns

```python
# Demographic columns of interest
demographic_columns = [col for col in clinical.columns if 'demographic' in col]
print("Demographic columns of interest:\n")
print("\n".join(demographic_columns))

# Create subset DataFrame
demo_df = clinical[demographic_columns].copy()
print(f"\nDemographic subset shape: {demo_df.shape}")
```

## Part 1: Data Quality Assessment

```python
# Calculate missing data statistics
missing_stats = pd.DataFrame({
    'column': demo_df.columns,
    'missing_count': demo_df.isnull().sum(),
    'missing_pct': (demo_df.isnull().sum() / len(demo_df)) * 100,
    'present_count': demo_df.notnull().sum(),
    'unique_values': [demo_df[col].nunique() for col in demo_df.columns]
})

missing_stats = missing_stats.sort_values('missing_pct', ascending=False)

print("Data Quality Summary:")
print("="*80)
print(f"Total patients: {len(demo_df):,}")
print(f"Total columns: {len(demo_df.columns)}")
print(f"\nColumns by completeness:")
print(
    f"  Complete (0% missing):        {len(missing_stats[missing_stats['missing_pct'] == 0]):3d}")
print(
    f"  High quality (<10% missing):  {len(missing_stats[missing_stats['missing_pct'] < 10]):3d}")
print(
    f"  Medium quality (10-50%):      {len(missing_stats[(missing_stats['missing_pct'] >= 10) & (missing_stats['missing_pct'] < 50)]):3d}")
print(
    f"  Low quality (>50% missing):   {len(missing_stats[missing_stats['missing_pct'] >= 50]):3d}")

print("\n" + "="*80)
print("All demographic columns:")
print(missing_stats.to_string(index=False))
```

```python
# HoloViews visualization: Data completeness bar chart
# Prepare data
sorted_stats = missing_stats.sort_values('missing_pct', ascending=True)

# Shorten column names for display
short_names = [col.replace('.demographic', '')
               for col in sorted_stats['column']]
sorted_stats['short_name'] = short_names

# Color map based on missing percentage


def get_color(pct):
    if pct == 0:
        return '#2ecc71'  # Green - complete
    elif pct < 10:
        return '#3498db'  # Blue - high quality
    elif pct < 50:
        return '#f39c12'  # Orange - medium
    else:
        return '#e74c3c'  # Red - low quality


sorted_stats['color'] = sorted_stats['missing_pct'].apply(get_color)

# Create HoloViews Bars
bars = hv.Bars(sorted_stats, kdims=['short_name'], vdims=[
               'missing_pct', 'color', 'column', 'missing_count', 'present_count', 'unique_values'])

bars.opts(
    opts.Bars(
        width=900,
        height=500,
        invert_axes=True,
        color='color',
        xlabel='Missing Data (%)',
        ylabel='Column',
        title='Missing Data Across All Demographic Columns',
        tools=['hover'],
        xrotation=0,
        show_legend=False,
        toolbar='right'
    )
)

show(hv.render(bars))
```

## Part 5: Age Analysis

```python
# Age at index
if age_col in demo_df.columns:
    age_data = pd.to_numeric(demo_df[age_col], errors='coerce').dropna()

    print("Age at Index Statistics:")
    print(f"  Count: {len(age_data):,}")
    print(f"  Mean: {age_data.mean():.1f} years")
    print(f"  Median: {age_data.median():.1f} years")
    print(f"  Std Dev: {age_data.std():.1f} years")
    print(f"  Range: {age_data.min():.0f} - {age_data.max():.0f} years")
    print(f"  Q1 (25%): {age_data.quantile(0.25):.1f} years")
    print(f"  Q3 (75%): {age_data.quantile(0.75):.1f} years")
    print(
        f"  IQR: {age_data.quantile(0.75) - age_data.quantile(0.25):.1f} years")
```

```python
# HoloViews: Age distribution with KDE overlay
if age_col in demo_df.columns:
    age_data = pd.to_numeric(demo_df[age_col], errors='coerce').dropna()

    # Create histogram
    hist = hv.Histogram(np.histogram(age_data, bins=40))

    # Create KDE curve
    from scipy import stats
    kde = stats.gaussian_kde(age_data)
    x_range = np.linspace(age_data.min(), age_data.max(), 200)
    kde_values = kde(x_range)

    # Normalize KDE to match histogram scale
    hist_counts, _ = np.histogram(age_data, bins=40)
    kde_scaled = kde_values * len(age_data) * \
        (age_data.max() - age_data.min()) / 40

    kde_curve = hv.Curve((x_range, kde_scaled), kdims=[
                         'Age'], vdims=['Density'])

    # Overlay histogram and KDE
    overlay = (hist * kde_curve).relabel('Age at Index Distribution')

    overlay.opts(
        opts.Histogram(
            width=900,
            height=500,
            fill_color='#3498db',
            fill_alpha=0.7,
            line_color='white',
            xlabel='Age (years)',
            ylabel='Count',
            tools=['hover']
        ),
        opts.Curve(
            color='#e74c3c',
            line_width=3,
            alpha=0.8
        )
    )

    show(hv.render(overlay))
```

```python
# Age groups analysis
if age_col in demo_df.columns:
    age_data = pd.to_numeric(demo_df[age_col], errors='coerce').dropna()

    age_groups = pd.cut(age_data, bins=[0, 40, 50, 60, 70, 100],
                        labels=['<40', '40-50', '50-60', '60-70', '70+'])
    age_group_counts = age_groups.value_counts().sort_index()

    print("\nAge Groups Distribution:")
    for group, count in age_group_counts.items():
        print(f"  {group}: {count:,} ({count/len(age_data)*100:.1f}%)")
```

```python
# HoloViews: Age groups bar chart
if age_col in demo_df.columns:
    age_data = pd.to_numeric(demo_df[age_col], errors='coerce').dropna()
    age_groups = pd.cut(age_data, bins=[0, 40, 50, 60, 70, 100],
                        labels=['<40', '40-50', '50-60', '60-70', '70+'])
    age_group_counts = age_groups.value_counts().sort_index()

    # Prepare data
    group_data = pd.DataFrame({
        'groups': [str(x) for x in age_group_counts.index],
        'counts': age_group_counts.values,
        'percentages': [f"{(c/len(age_data)*100):.1f}%" for c in age_group_counts.values]
    })

    bars = hv.Bars(group_data, kdims=['groups'], vdims=[
                   'counts', 'percentages'])

    bars.opts(
        opts.Bars(
            width=900,
            height=500,
            color=hv.Cycle(
                ['#3498db', '#e74c3c', '#2ecc71', '#f39c12', '#9b59b6']),
            xlabel='Age Group',
            ylabel='Number of Patients',
            title='Age Groups Distribution',
            tools=['hover'],
            show_legend=False,
            toolbar='right',
            alpha=0.8,
            line_color='white'
        )
    )

    show(hv.render(bars))
```

```python
# Days to birth analysis (represents age in days)
days_birth_col = 'days_to_birth.demographic'

if days_birth_col in demo_df.columns:
    days_birth = pd.to_numeric(
        demo_df[days_birth_col], errors='coerce').dropna()

    # Convert to years (negative values, so multiply by -1)
    age_from_days = (days_birth * -1) / 365.25

    print("Days to Birth (as Age) Statistics:")
    print(f"  Count: {len(age_from_days):,}")
    print(f"  Mean: {age_from_days.mean():.1f} years")
    print(f"  Median: {age_from_days.median():.1f} years")
    print(
        f"  Range: {age_from_days.min():.1f} - {age_from_days.max():.1f} years")
```

```python
# Year of birth analysis
yob_col = 'year_of_birth.demographic'

if yob_col in demo_df.columns:
    yob_data = pd.to_numeric(demo_df[yob_col], errors='coerce').dropna()

    print("\nYear of Birth Statistics:")
    print(f"  Count: {len(yob_data):,}")
    print(f"  Range: {yob_data.min():.0f} - {yob_data.max():.0f}")
    print(f"  Median: {yob_data.median():.0f}")
    print(f"  Mode: {yob_data.mode().values[0]:.0f}" if len(
        yob_data.mode()) > 0 else "  Mode: N/A")

    # Decade analysis
    decades = (yob_data // 10 * 10).value_counts().sort_index()
    print("\nBirth Decade Distribution:")
    for decade, count in decades.items():
        print(f"  {decade:.0f}s: {count:,} ({count/len(yob_data)*100:.1f}%)")
```

```python
# HoloViews: Year of birth distribution
if yob_col in demo_df.columns:
    yob_data = pd.to_numeric(demo_df[yob_col], errors='coerce').dropna()

    # Create histogram
    hist = hv.Histogram(np.histogram(yob_data, bins=50))

    # Add median line
    median_line = hv.VLine(yob_data.median()).opts(
        color='red',
        line_width=2,
        line_dash='dashed'
    )

    overlay = (hist * median_line).relabel(f'Year of Birth Distribution')

    overlay.opts(
        opts.Histogram(
            width=900,
            height=500,
            fill_color='#9b59b6',
            fill_alpha=0.7,
            line_color='white',
            xlabel='Year of Birth',
            ylabel='Number of Patients',
            tools=['hover']
        )
    )

    show(hv.render(overlay))
```

```python
# Year of death analysis
if yod_col in demo_df.columns:
    yod_data = pd.to_numeric(demo_df[yod_col], errors='coerce').dropna()

    print("Year of Death Statistics:")
    print(f"  Count: {len(yod_data):,}")
    print(f"  Missing: {demo_df[yod_col].isnull().sum():,}")
    print(f"  % of cohort: {len(yod_data)/len(demo_df)*100:.1f}%")

    if len(yod_data) > 0:
        print(f"  Range: {yod_data.min():.0f} - {yod_data.max():.0f}")
        print(f"  Median: {yod_data.median():.0f}")
        print(f"  Mode: {yod_data.mode().values[0]:.0f}" if len(
            yod_data.mode()) > 0 else "  Mode: N/A")
```

```python
# HoloViews: Year of death distribution
if yod_col in demo_df.columns:
    yod_data = pd.to_numeric(demo_df[yod_col], errors='coerce').dropna()

    if len(yod_data) > 0:
        # Create histogram
        hist = hv.Histogram(np.histogram(yod_data, bins=30))

        # Add median line
        median_line = hv.VLine(yod_data.median()).opts(
            color='darkblue',
            line_width=2,
            line_dash='dashed'
        )

        overlay = (
            hist * median_line).relabel(f'Year of Death Distribution (n={len(yod_data):,})')

        overlay.opts(
            opts.Histogram(
                width=900,
                height=500,
                fill_color='#c0392b',
                fill_alpha=0.7,
                line_color='white',
                xlabel='Year of Death',
                ylabel='Number of Patients',
                tools=['hover']
            )
        )

        show(hv.render(overlay))
    else:
        print("No year of death data available for visualization")
```

```python
# Days to death analysis
dtd_col = 'days_to_death.demographic'

if dtd_col in demo_df.columns:
    dtd_data = pd.to_numeric(demo_df[dtd_col], errors='coerce').dropna()

    print("\nDays to Death Statistics:")
    print(f"  Count: {len(dtd_data):,}")
    print(f"  Missing: {demo_df[dtd_col].isnull().sum():,}")

    if len(dtd_data) > 0:
        dtd_years = dtd_data / 365.25
        print(f"  Mean: {dtd_years.mean():.1f} years")
        print(f"  Median: {dtd_years.median():.1f} years")
        print(f"  Range: {dtd_years.min():.1f} - {dtd_years.max():.1f} years")
        print(f"  Q1: {dtd_years.quantile(0.25):.1f} years")
        print(f"  Q3: {dtd_years.quantile(0.75):.1f} years")
```

```python
# HoloViews: Days to death distribution
if dtd_col in demo_df.columns:
    dtd_data = pd.to_numeric(demo_df[dtd_col], errors='coerce').dropna()

    if len(dtd_data) > 0:
        dtd_years = dtd_data / 365.25

        # Create histogram
        hist = hv.Histogram(np.histogram(dtd_years, bins=40))

        # Add median and mean lines
        median_line = hv.VLine(dtd_years.median()).opts(
            color='darkblue',
            line_width=2,
            line_dash='dashed'
        )

        mean_line = hv.VLine(dtd_years.mean()).opts(
            color='orange',
            line_width=2,
            line_dash='dashed'
        )

        overlay = (hist * median_line * mean_line).relabel(
            f'Time to Death Distribution (n={len(dtd_data):,})')

        overlay.opts(
            opts.Histogram(
                width=900,
                height=500,
                fill_color='#e74c3c',
                fill_alpha=0.7,
                line_color='white',
                xlabel='Time to Death (years)',
                ylabel='Number of Patients',
                tools=['hover']
            )
        )

        show(hv.render(overlay))
    else:
        print("No days to death data available for visualization")
```

## Part 7: Cross-Demographic Comparisons

```python
# Age by Gender
if age_col in demo_df.columns and gender_col in demo_df.columns:
    # Setup
    age_gender_df = demo_df[[age_col, gender_col]].copy()
    age_gender_df['age'] = pd.to_numeric(
        age_gender_df[age_col], errors='coerce')
    age_gender_df = age_gender_df.dropna()

    # Statistics
    print("Age Statistics by Gender:")
    for gender in age_gender_df[gender_col].unique():
        gender_ages = age_gender_df[age_gender_df[gender_col] == gender]['age']
        print(f"\n{gender}:")
        print(f"  Count: {len(gender_ages):,}")
        print(f"  Mean: {gender_ages.mean():.1f} years")
        print(f"  Median: {gender_ages.median():.1f} years")
        print(f"  Std: {gender_ages.std():.1f} years")

    # HoloViews: Overlaid histograms by gender
    genders = sorted(age_gender_df[gender_col].unique())
    colors = ['#3498db', '#e74c3c', '#2ecc71']

    overlays = []
    for i, gender in enumerate(genders):
        gender_ages = age_gender_df[age_gender_df[gender_col] == gender]['age']

        if len(gender_ages) > 10:  # Only plot if sufficient data
            hist = hv.Histogram(np.histogram(
                gender_ages, bins=30, density=True))
            hist = hist.relabel(f"{gender} (n={len(gender_ages)})")
            hist = hist.opts(
                fill_color=colors[i % len(colors)],
                fill_alpha=0.5,
                line_color='white',
                width=900,
                height=500,
                xlabel='Age (years)',
                ylabel='Density',
                tools=['hover']
            )
            overlays.append(hist)

    if overlays:
        combined = hv.Overlay(overlays).relabel('Age Distribution by Gender')
        combined.opts(legend_position='top_right', toolbar='right')

        show(hv.render(combined))
```

```python
# Age by Race
if age_col in demo_df.columns and race_col in demo_df.columns:
    # Setup
    age_race_df = demo_df[[age_col, race_col]].copy()
    age_race_df['age'] = pd.to_numeric(age_race_df[age_col], errors='coerce')
    age_race_df = age_race_df.dropna()

    races = age_race_df[race_col].value_counts().index

    # Statistics
    print("\nAge Statistics by Race:")
    for race in races:
        race_ages = age_race_df[age_race_df[race_col] == race]['age']
        print(f"\n{race}:")
        print(f"  Count: {len(race_ages):,}")
        print(f"  Mean: {race_ages.mean():.1f} years")
        print(f"  Median: {race_ages.median():.1f} years")
        print(f"  Q1 (25th percentile): {race_ages.quantile(0.25):.1f} years")
        print(f"  Q2 (50th percentile): {race_ages.quantile(0.50):.1f} years")
        print(f"  Q3 (75th percentile): {race_ages.quantile(0.75):.1f} years")
        print(f"  Q4 (Max): {race_ages.max():.1f} years")

    # HoloViews BoxWhisker plot
    # Prepare data for HoloViews BoxWhisker
    box_data = []
    for race in races:
        race_ages = age_race_df[age_race_df[race_col] == race]['age']
        for age in race_ages:
            box_data.append((race, age))

    # Create BoxWhisker plot
    boxwhisker = hv.BoxWhisker(box_data, kdims=['Race'], vdims=['Age'])

    # Configure the plot
    boxwhisker.opts(
        opts.BoxWhisker(
            width=900,
            height=600,
            box_fill_color='#3498db',
            box_alpha=0.7,
            box_line_color='#2c3e50',
            whisker_color='#2c3e50',
            outlier_color='#e74c3c',
            title=f"Age Distribution by Race - Box Plot with Quartiles (n={len(age_race_df):,})",
            xlabel='Race',
            ylabel='Age (years)',
            xrotation=45,
            show_legend=False,
            tools=['hover', 'pan', 'wheel_zoom', 'box_zoom', 'reset', 'save']
        )
    )

    # Display the plot
    show(hv.render(boxwhisker))
```

```python
# Vital Status by Gender
if vital_col in demo_df.columns and gender_col in demo_df.columns:
    vital_gender_df = demo_df[[vital_col, gender_col]].dropna()

    crosstab = pd.crosstab(
        vital_gender_df[gender_col], vital_gender_df[vital_col])

    print("\nVital Status by Gender:")
    print(crosstab)
    print("\nPercentages by Gender:")
    print(crosstab.div(crosstab.sum(axis=1), axis=0) * 100)
```

```python
# HoloViews: Grouped bar chart - Vital Status by Gender
if vital_col in demo_df.columns and gender_col in demo_df.columns:
    vital_gender_df = demo_df[[vital_col, gender_col]].dropna()
    crosstab = pd.crosstab(
        vital_gender_df[gender_col], vital_gender_df[vital_col])

    # Prepare data for HoloViews
    data_list = []
    for gender in crosstab.index:
        for status in crosstab.columns:
            data_list.append((gender, status, crosstab.loc[gender, status]))

    bars = hv.Bars(data_list, kdims=['Gender', 'Status'], vdims=['Count'])

    # Define colors for statuses
    color_map = {'Alive': '#2ecc71',
                 'Dead': '#e74c3c', 'Not Reported': '#95a5a6'}

    bars.opts(
        opts.Bars(
            width=900,
            height=500,
            color=dim('Status').categorize(color_map, default='#3498db'),
            xlabel='Gender',
            ylabel='Number of Patients',
            title='Vital Status by Gender',
            tools=['hover'],
            legend_position='top_right',
            toolbar='right',
            alpha=0.8,
            show_legend=True
        )
    )

    show(hv.render(bars))
```

```python
# Create comprehensive summary table
summary_data = []

for col in existing_cols:
    col_data = demo_df[col]

    # Basic stats
    total = len(col_data)
    missing = col_data.isnull().sum()
    present = total - missing
    unique = col_data.nunique()

    # Try to get numeric stats if applicable
    numeric_data = pd.to_numeric(col_data, errors='coerce')
    if numeric_data.notnull().sum() > 0:
        mean_val = numeric_data.mean()
        median_val = numeric_data.median()
        std_val = numeric_data.std()
        data_type = 'Numeric'
    else:
        mean_val = None
        median_val = None
        std_val = None
        data_type = 'Categorical'

    # Most common value
    if present > 0:
        mode_val = col_data.mode().values[0] if len(
            col_data.mode()) > 0 else None
        mode_count = col_data.value_counts().iloc[0] if len(
            col_data.value_counts()) > 0 else 0
    else:
        mode_val = None
        mode_count = 0

    summary_data.append({
        'Column': col.replace('.demographic', ''),
        'Type': data_type,
        'Total': total,
        'Present': present,
        'Missing': missing,
        'Missing %': f"{missing/total*100:.1f}",
        'Unique': unique,
        'Mode': str(mode_val)[:30] if mode_val else 'N/A',
        'Mode Count': mode_count,
        'Mean': f"{mean_val:.2f}" if mean_val is not None else 'N/A',
        'Median': f"{median_val:.2f}" if median_val is not None else 'N/A',
        'Std Dev': f"{std_val:.2f}" if std_val is not None else 'N/A'
    })

summary_df = pd.DataFrame(summary_data)

print("\nComprehensive Demographic Summary Statistics:")
print("="*120)
print(summary_df.to_string(index=False))
```
